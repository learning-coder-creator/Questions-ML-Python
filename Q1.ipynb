```python
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# 1. Load dataset
# For custom CSV files, replace with: df = pd.read_csv("data.csv")
iris = load_iris(as_frame=True)
df = iris.frame

X = df.drop(columns=["target"])
y = df["target"]

# 2. Perform 70/30 train/test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42
)

# 3. Apply standard scaling (StandardScaler)
scaler = StandardScaler()

# Fit on training set only, then transform both to prevent data leakage
X_train_scaled = pd.DataFrame(
    scaler.fit_transform(X_train), columns=X_train.columns, index=X_train.index
)

X_test_scaled = pd.DataFrame(
    scaler.transform(X_test), columns=X_test.columns, index=X_test.index
)

# Print set sizes to verify 70/30 split ratio
print(f"Training features shape: {X_train_scaled.shape}")
print(f"Testing features shape:  {X_test_scaled.shape}")
```
